## Matrices/Arrays and broadcasting via dot

This is an example of exploration for matrices as a natural operation within the default environment of `Julia` without any need to add and use any package.

```julia
M = [ 1 2 3 
      4 5 6 
      7 8 9 ]; # create a 3x3 matrix
println(M); # print the matrix to the console

# Random-valued matrix within the range of 0 to 1
M = rand(3,3); # create a 3x3 matrix with random values

println("sin.(M) = ",sin.(M)); # print the sine of each element in the matrix to the console
println("cos.(M) = ",cos.(M)); # print the cosine of each element in the matrix to the console
println("tan.(M) = ",tan.(M)); # print the tangent of each element in the matrix to the console
```
